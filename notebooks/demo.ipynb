# Closed-loop UUV depth control

Simulates the UUV following the mission reference using the PD controller from `uuv_mission.control`.

```python
import sys  
sys.path.append("..") # repo root, so uuv_mission can be imported
from uuv_mission.dynamic import Submarine, ClosedLoop, Mission 
from uuv_mission.control import PDController
```

## Load the mission and build the closed loop

```python
mission = Mission.from_csv("../data/mission.csv")
controller = PDController()
closed_loop = ClosedLoop(Submarine(), controller)
```

## Simulate and plot

Random disturbances are applied, so the trajectory differs slightly on each run.

```python
trajectory = closed_loop.simulate_with_random_disturbances(mission)
trajectory.plot_completed_mission(mission)
```
